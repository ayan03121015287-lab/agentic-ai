# Multiple Choices / Conditions

```python
percentage = float(input("Enter your percentage: "))
if percentage>=90:
    print("A+")
elif percentage>=80:
    print("A")
elif percentage>=70:
    print("B")
elif percentage>=60:
    print("C")
elif percentage>=50:
    print("D")
else:
    print("F")
```
